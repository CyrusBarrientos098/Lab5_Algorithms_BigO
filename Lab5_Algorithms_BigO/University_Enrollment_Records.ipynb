```python
LAST_NAME = "Barrientos"
STUDENT_ID = "TUPM-26-4184"
FAVORITE_ARTIST = "Mac DeMarco"

digits = [int(ch) for ch in STUDENT_ID if ch.isdigit()]
SEED_DIGIT = digits[-1]
ID_SUM = sum(digits)
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(ch.upper()) - ord("A") + 1 for ch in FAVORITE_ARTIST if ch.isalpha())

print("LAST_NAME:", LAST_NAME)
print("STUDENT_ID:", STUDENT_ID)
print("FAVORITE_ARTIST:", FAVORITE_ARTIST)
print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("NAME_LENGTH:", NAME_LENGTH)
print("ARTIST_LENGTH:", ARTIST_LENGTH)
print("ARTIST_SUM:", ARTIST_SUM)

size_a, size_b = 8, 7
department_a = [100000 + ID_SUM * 100 + SEED_DIGIT * 10 + i * (SEED_DIGIT + 2) for i in range(size_a)]
department_b = [200000 + ARTIST_LENGTH * 100 + NAME_LENGTH * 10 + i * (NAME_LENGTH + 1) for i in range(size_b)]
department_a.sort()
department_b.sort()
print("Department A:", department_a)
print("Department B:", department_b)


def merge_sorted(a, b):
    i = j = comparisons = 0
    merged = []
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i]); i += 1
        else:
            merged.append(b[j]); j += 1
    merged.extend(a[i:])
    merged.extend(b[j:])
    return merged, comparisons

consolidated, merge_comparisons = merge_sorted(department_a, department_b)
is_sorted = all(consolidated[i] <= consolidated[i+1] for i in range(len(consolidated)-1))
print("Consolidated dataset:", consolidated)
print("Records before:", len(department_a), "+", len(department_b), "=", len(department_a)+len(department_b))
print("Records after:", len(consolidated))
print("Merge comparisons:", merge_comparisons)
print("Ordering verified:", is_sorted)


print("Merge time complexity: O(n + m), where n and m are dataset sizes.")
print("Auxiliary space complexity: O(n + m) for the merged output list.")


print({
    "department_a": department_a,
    "department_b": department_b,
    "consolidated_dataset": consolidated,
    "records_before": len(department_a) + len(department_b),
    "records_after": len(consolidated),
    "merge_comparisons": merge_comparisons,
    "ordering_verified": is_sorted
})
```
