```python

import time

# ==========================================
# LAB 5: ALGORITHMS AND BIG O
# APPENDIX: E-COMMERCE ORDER PROCESSING
# ==========================================

# 1. STUDENT INPUTS
LAST_NAME = "Barrientos"
STUDENT_ID = "TUPM-26-4184"
FAVORITE_ARTIST = "Mac DeMarco"

digits = [int(ch) for ch in STUDENT_ID if ch.isdigit()]

SEED_DIGIT = digits[-1]
ID_SUM = sum(digits)
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(
    ord(ch.upper()) - ord("A") + 1
    for ch in FAVORITE_ARTIST
    if ch.isalpha()
)

print("=" * 60)
print("STUDENT INFORMATION")
print("=" * 60)
print("LAST_NAME:", LAST_NAME)
print("STUDENT_ID:", STUDENT_ID)
print("FAVORITE_ARTIST:", FAVORITE_ARTIST)
print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("NAME_LENGTH:", NAME_LENGTH)
print("ARTIST_LENGTH:", ARTIST_LENGTH)
print("ARTIST_SUM:", ARTIST_SUM)


# ==========================================
# 2. GENERATE ORDER DATASET
# ==========================================

order_count = 12

base_orders = sorted(set(
    (
        ID_SUM * 10
        + ARTIST_SUM
        + i * (NAME_LENGTH + SEED_DIGIT + 1)
    ) % 997
    for i in range(order_count)
))

# Ensure the dataset contains 12 values
while len(base_orders) < order_count:
    next_value = base_orders[-1] + NAME_LENGTH + 1
    base_orders.append(next_value)

base_orders = sorted(base_orders)

sorted_input = base_orders.copy()
reverse_input = list(reversed(base_orders))
random_input = base_orders.copy()

# Deterministic shuffle based on student inputs
for i in range(len(random_input) - 1, 0, -1):
    j = (ID_SUM + ARTIST_SUM + i * SEED_DIGIT) % (i + 1)
    random_input[i], random_input[j] = (
        random_input[j],
        random_input[i]
    )

inputs = {
    "Sorted": sorted_input,
    "Reverse-Sorted": reverse_input,
    "Random": random_input
}

print("\n" + "=" * 60)
print("GENERATED ORDER DATASET")
print("=" * 60)
print("Base orders:", base_orders)

for condition, values in inputs.items():
    print(f"{condition} input: {values}")


# ==========================================
# 3. SORTING ALGORITHMS
# ==========================================

def bubble_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for end in range(len(a) - 1, 0, -1):
        swapped = False

        for i in range(end):
            comparisons += 1

            if a[i] > a[i + 1]:
                a[i], a[i + 1] = a[i + 1], a[i]
                movements += 3  # Three assignments per swap
                swapped = True

        if not swapped:
            break

    return a, comparisons, movements


def selection_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(a) - 1):
        min_index = i

        for j in range(i + 1, len(a)):
            comparisons += 1

            if a[j] < a[min_index]:
                min_index = j

        if min_index != i:
            a[i], a[min_index] = a[min_index], a[i]
            movements += 3  # Three assignments per swap

    return a, comparisons, movements


def insertion_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if a[j] > key:
                a[j + 1] = a[j]
                movements += 1  # One shift
                j -= 1
            else:
                break

        a[j + 1] = key
        movements += 1  # Count key placement

    return a, comparisons, movements


algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}


# ==========================================
# 4. TEST AND MEASURE EACH ALGORITHM
# ==========================================

results = {}

print("\n" + "=" * 60)
print("SORTING RESULTS")
print("=" * 60)

for condition, values in inputs.items():
    results[condition] = {}

    print(f"\n--- {condition} Input ---")

    for name, algorithm in algorithms.items():

        start_time = time.perf_counter()

        output, comparisons, movements = algorithm(values)

        execution_time = time.perf_counter() - start_time

        correctly_sorted = output == sorted(values)

        results[condition][name] = {
            "sorted_output": output,
            "comparisons": comparisons,
            "movements": movements,
            "execution_time_seconds": execution_time,
            "correctly_sorted": correctly_sorted
        }

        print(f"\nAlgorithm: {name}")
        print("Sorted output:", output)
        print("Comparison count:", comparisons)
        print("Movement count:", movements)
        print(f"Execution time: {execution_time:.10f} seconds")
        print("Sorting verified:", correctly_sorted)


# ==========================================
# 5. BIG O ANALYSIS
# ==========================================

print("\n" + "=" * 60)
print("BIG O COMPLEXITY ANALYSIS")
print("=" * 60)

complexities = {
    "Bubble Sort": {
        "Best Case": "O(n)",
        "Average Case": "O(n^2)",
        "Worst Case": "O(n^2)",
        "Auxiliary Space": "O(1)"
    },
    "Selection Sort": {
        "Best Case": "O(n^2)",
        "Average Case": "O(n^2)",
        "Worst Case": "O(n^2)",
        "Auxiliary Space": "O(1)"
    },
    "Insertion Sort": {
        "Best Case": "O(n)",
        "Average Case": "O(n^2)",
        "Worst Case": "O(n^2)",
        "Auxiliary Space": "O(1)"
    }
}

for algorithm, analysis in complexities.items():
    print(f"\n{algorithm}")
    for case, complexity in analysis.items():
        print(f"{case}: {complexity}")

print(
    "\nNote: O(1) auxiliary space describes the sorting logic. "
    "The implementation also copies the input list, which uses O(n) space."
)


# ==========================================
# 6. BEST-, AVERAGE-, AND WORST-CASE OBSERVATIONS
# ==========================================

print("\n" + "=" * 60)
print("CASE OBSERVATIONS")
print("=" * 60)

print(
    "Best case: Sorted input can reduce the work performed by "
    "Bubble Sort and Insertion Sort."
)
print(
    "Average case: Random input illustrates typical behavior, "
    "but one random arrangement does not establish the true average."
)
print(
    "Worst case: Reverse-sorted input generally causes more work "
    "for Bubble Sort and Insertion Sort."
)
print(
    "Selection Sort performs the same number of comparisons "
    "for these input arrangements of equal length."
)

for condition, group in results.items():
    print(f"\n{condition} input observations:")

    for name, item in group.items():
        print(
            f"{name}: "
            f"{item['comparisons']} comparisons, "
            f"{item['movements']} movements, "
            f"{item['execution_time_seconds']:.10f} seconds"
        )


# ==========================================
# 7. SORTING VERIFICATION SUMMARY
# ==========================================

print("\n" + "=" * 60)
print("SORTING VERIFICATION")
print("=" * 60)

all_verified = True

for condition, group in results.items():
    for name, item in group.items():
        status = item["correctly_sorted"]
        print(f"{condition} | {name}: {status}")

        if not status:
            all_verified = False

print("\nAll algorithms correctly sorted their inputs:", all_verified)


# ==========================================
# 8. FINAL OUTPUT
# ==========================================

final_output = {
    "student_inputs": {
        "LAST_NAME": LAST_NAME,
        "STUDENT_ID": STUDENT_ID,
        "FAVORITE_ARTIST": FAVORITE_ARTIST,
        "SEED_DIGIT": SEED_DIGIT,
        "ID_SUM": ID_SUM,
        "NAME_LENGTH": NAME_LENGTH,
        "ARTIST_LENGTH": ARTIST_LENGTH,
        "ARTIST_SUM": ARTIST_SUM
    },
    "dataset": base_orders,
    "input_conditions": inputs,
    "results": results,
    "big_o_analysis": complexities,
    "all_sorting_verified": all_verified
}

print("\n" + "=" * 60)
print("FINAL OUTPUT")
print("=" * 60)
print(final_output)

```
